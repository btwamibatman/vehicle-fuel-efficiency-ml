# Engine characteristic clusters

Train-only K-Means uses cylinder count and engine displacement (L). Median imputation and standardization are fit on training rows. MPG is not used to fit or choose clusters.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Image
root = Path.cwd()
while not (root / 'pyproject.toml').exists():
    if root == root.parent: raise RuntimeError('Run within the project')
    root = root.parent
r = root / 'reports/results/endterm'
f = root / 'reports/figures/endterm'
```

```python
display(pd.read_csv(r / 'clustering_search.csv'))
display(Image(filename=str(f / 'clustering_inertia.png')))
display(Image(filename=str(f / 'clustering_silhouette.png')))
```

Three clusters give a substantial reduction in inertia compared with two and a strong silhouette. Four has a slightly higher silhouette but creates a smaller group; three is a compact, interpretable partition, not a claim that precisely three natural groups exist. Silhouette uses a fixed 3,000-row sample. Scaling prevents the numerical range of either input from dominating the distance.

```python
display(pd.read_csv(r / 'clustering_profiles.csv'))
display(Image(filename=str(f / 'clustering_scatter.png')))
```

Centroids are transformed back to original units and represent average imputed feature values. IDs are arbitrary labels, not rankings. MPG is summarized only after clustering; differences are descriptive and do not establish causation. The clustering describes engines, not all vehicle properties.
